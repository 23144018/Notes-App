```python
query1 = """
SELECT
    c.Country,
    strftime('%Y-%m', i.InvoiceDate) AS Month,
    SUM(il.Quantity * il.UnitPrice) AS Revenue
FROM invoices i
JOIN customers c
    ON i.CustomerID = c.CustomerID
JOIN invoice_lines il
    ON i.InvoiceNo = il.InvoiceNo
GROUP BY
    c.Country,
    Month
ORDER BY
    c.Country,
    Month;
"""

start = time.perf_counter()

cursor.execute(query1)
result1 = cursor.fetchall()

end = time.perf_counter()

time1 = (end - start) * 1000

print("Query 1 completed successfully!")
print("Number of result rows:", len(result1))
print("Time:", time1, "ms")

print("\nFirst 10 results:")
for row in result1[:10]:
    print(row)
```

```python
query2 = """
SELECT
    p.StockCode,
    p.Description,
    SUM(il.Quantity * il.UnitPrice) AS Revenue
FROM invoice_lines il
JOIN products p
    ON il.StockCode = p.StockCode
GROUP BY
    p.StockCode,
    p.Description
ORDER BY
    Revenue DESC
LIMIT 10;
"""

start = time.perf_counter()

cursor.execute(query2)
result2 = cursor.fetchall()

end = time.perf_counter()

time2 = (end - start) * 1000

print("Query 2 completed successfully!")
print("Time:", time2, "ms")

print("\nTop 10 products:")
for row in result2:
    print(row)
```

```python
query3 = """
SELECT
    100.0 *
    SUM(
        CASE
            WHEN invoice_count > 1 THEN 1
            ELSE 0
        END
    ) / COUNT(*) AS repeat_customer_rate
FROM (
    SELECT
        CustomerID,
        COUNT(DISTINCT InvoiceNo) AS invoice_count
    FROM invoices
    WHERE CustomerID IS NOT NULL
    GROUP BY CustomerID
);
"""

start = time.perf_counter()

cursor.execute(query3)
result3 = cursor.fetchone()

end = time.perf_counter()

time3 = (end - start) * 1000

print("Query 3 completed successfully!")
print("Repeat customer rate:", result3[0], "%")
print("Time:", time3, "ms")
```

```python
dss_sql = """
-- Q1: Revenue per country per month

SELECT
    c.Country,
    strftime('%Y-%m', i.InvoiceDate) AS Month,
    SUM(il.Quantity * il.UnitPrice) AS Revenue
FROM invoices i
JOIN customers c
    ON i.CustomerID = c.CustomerID
JOIN invoice_lines il
    ON i.InvoiceNo = il.InvoiceNo
GROUP BY
    c.Country,
    Month
ORDER BY
    c.Country,
    Month;


-- Q2: Top 10 products by revenue

SELECT
    p.StockCode,
    p.Description,
    SUM(il.Quantity * il.UnitPrice) AS Revenue
FROM invoice_lines il
JOIN products p
    ON il.StockCode = p.StockCode
GROUP BY
    p.StockCode,
    p.Description
ORDER BY
    Revenue DESC
LIMIT 10;


-- Q3: Repeat-customer rate

SELECT
    100.0 *
    SUM(
        CASE
            WHEN invoice_count > 1 THEN 1
            ELSE 0
        END
    ) / COUNT(*) AS repeat_customer_rate
FROM (
    SELECT
        CustomerID,
        COUNT(DISTINCT InvoiceNo) AS invoice_count
    FROM invoices
    WHERE CustomerID IS NOT NULL
    GROUP BY CustomerID
);
"""

with open("/content/DMW_Lab/Unit1/u1_dss_queries.sql", "w") as f:
    f.write(dss_sql)

print("u1_dss_queries.sql created successfully!")
```
