```python
import os
import pandas as pd
import numpy as np
import sqlite3
import time

os.makedirs("/DMW_Lab/Unit1", exist_ok=True)

print("Folder created")
```

```python
url = "https://archive.ics.uci.edu/ml/machine-learning-databases/00352/Online%20Retail.xlsx"

df = pd.read_excel(url)

print(df.shape)
df.head()
```

```python
print("Shape:", df.shape)

print("\nData types:")
print(df.dtypes)

print("\nNull counts:")
print(df.isnull().sum())

print("\nDistinct counts:")
print(df.nunique())
```

```python
profile = pd.DataFrame({
    "column": df.columns,
    "dtype": [str(df[col].dtype) for col in df.columns],
    "null_count": [df[col].isna().sum() for col in df.columns],
    "distinct_count": [df[col].nunique(dropna=True) for col in df.columns]
})

profile
```

```python
profile.to_csv(
    "/DMW_Lab/Unit1/u1_profile.csv",
    index=False
)
```
